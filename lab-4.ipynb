Problem 1.
Load the Titanic dataset using Pandas. Display the first five rows and inspect the dataset. Identify the
numerical and categorical attributes that can be used for visualization.

```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("c://Users//student//Downloads//train.csv")
df.head()
```

```python
# inspecting the dataset
print("shape:", df.shape)
print("columns:", df.columns)
print("dataset information:", df.info())
print("statistical summary:", df.describe())
```

Problem 2.
Create a line plot to visualize the average fare paid by passengers across different passenger classes.
Include an appropriate title and axis labels.


```python
avg_fare = df.groupby('Pclass')['Fare'].mean()
print(avg_fare)
plt.figure(figsize=(8, 5))

plt.plot(avg_fare.index, avg_fare.values)

plt.title('Average Fare Paid Across Passenger Classes')
plt.xlabel('Passenger Class')
plt.ylabel('Average Fare')

plt.xticks([1, 2, 3])
plt.grid(True)

plt.show()
```

Class 1 paid the highest average fare, while class 3 paid the lowest average fare

Problem 3.
Create a scatter plot to visualize the relationship between passenger age and fare. Add an appropriate
title and axis labels.

```python
plt.figure(figsize=(8, 5))

plt.scatter(df['Age'], df['Fare'])

plt.title('Relationship Between Passenger Age and Fare')
plt.xlabel('Passenger Age')
plt.ylabel('Fare')

plt.grid(True)

plt.show()
```

few passengers paid exceptionally high fares, with some fares exceeding 500. The presence of extreme fare values indicates some outliers in the dataset.

Problem 4.
Create a bar chart to compare the number of male and female passengers in the Titanic dataset. Include
an appropriate title and axis labels.


```python
gender_count = df['Sex'].value_counts()
print(gender_count)
plt.figure(figsize=(8, 5))
plt.bar(gender_count.index, gender_count.values)

plt.title('Number of Male and Female Passengers on Titanic')
plt.xlabel('Gender')
plt.ylabel('Number of Passengers')

plt.show()
```

male passengers significantly outnumbered female passengers. More than 500 passengers were male, and around 300 passengers were female

Problem 5.
Create a histogram to visualize the distribution of passenger ages. Use an appropriate number of bins
and add a title and axis labels

```python
plt.figure(figsize=(8, 5))

plt.hist(df['Age'])

plt.title('Distribution of Passenger Ages on Titanic')
plt.xlabel('Passenger Age')
plt.ylabel('Number of Passengers')

plt.grid(axis='y', alpha=0.3)

plt.show()
```

There were relatively few passengers above 60 years old. Most passengers were young adults

Problem 6.
Create a pie chart to represent the proportion of passengers who survived and did not survive. Label the
categories and display their percentages.

```python
survival_count = df['Survived'].value_counts()
print(survival_count)

plt.figure(figsize=(7, 7))

plt.pie(
    survival_count,
    labels=['Not Survived', 'Survived']
)

plt.title('Proportion of Survived and Not Survived Passengers')

plt.show()
```

more than half of the passengers did not survive

Problem 7.
Create a figure containing two subplots:
1. A histogram showing the distribution of passenger ages.
2. A bar chart showing the number of passengers in each passenger class.
Add appropriate titles to both subplots.


```python
plt.figure(figsize=(12, 5))

# Histogram of passenger ages
plt.subplot(1, 2, 1)
plt.hist(df['Age'])

plt.title('Distribution of Passenger Ages')
plt.xlabel('Age')
plt.ylabel('Number of Passengers')


# Bar chart of passenger classes
plt.subplot(1, 2, 2)
class_count = df['Pclass'].value_counts().sort_index()
plt.bar(class_count.index, class_count.values)
plt.title('Number of Passengers in Each Class')
plt.xlabel('Passenger Class')
plt.ylabel('Number of Passengers')


plt.show()
```

the 1st subplot shows distribution of passenger ages and the 2nd subplot shows the number of passengers in each class

Problem 8.
Create a figure containing three subplots to visualize:
1. The number of passengers by gender using a bar chart.
2. The distribution of passenger fares using a histogram.
3. The survival proportion using a pie chart.

```python
plt.figure(figsize=(15, 5))

# Number of passengers by gender using bar chart
plt.subplot(1, 3, 1)
gender_count = df['Sex'].value_counts()

plt.bar(gender_count.index, gender_count.values)
plt.title('Number of Passengers by Gender')
plt.xlabel('Gender')
plt.ylabel('Number of Passengers')


# Distribution of passenger fares using histogram
plt.subplot(1, 3, 2)
plt.hist(df['Fare'])
plt.title('Distribution of Passenger Fares')
plt.xlabel('Fare')
plt.ylabel('Frequency')


# Survival proportion using pie chart
plt.subplot(1, 3, 3)
survival_count = df['Survived'].value_counts().sort_index()
plt.pie(survival_count,
        labels=['Not Survived', 'Survived'])
plt.title('Survival Proportion')

plt.show()
```

1st subplot shows the number of male and female passengers, 2nd subplot shows the distribution of passenger fares, the 3rd subplot shows how many passangers survived and did not survive
