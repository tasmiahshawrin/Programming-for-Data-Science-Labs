1. Write a program that inputs two numbers and displays their sum, difference, product, and quotient. Convert the inputs to numeric values before calculating.

```python
a = float(input("Enter the first number: "))
b = float(input("Enter the second number: "))
print("Sum:", a + b)
print("Difference:", a - b)
print("Product:", a * b)
if b != 0:
    print("Quotient:", a / b)
else:
    print("Invalid")
```

3. Input a temperature in Celsius and convert it to Fahrenheit and Kelvin using the formulas:
F = (C × 9/5) + 32, K = C + 273.15

```python
c = float(input("Enter temperature in Celsius: "))
f = (c * 9 / 5) + 32
k = c + 273.15

print("Temperature in Celsius: ", c, "°C")
print("Temperature in Fahrenheit: ", f, "°F")
print("Temperature in Kelvin: ", k, "K")
```

4. Input a student’s name and five course marks. Calculate the total, average, highest mark, lowest mark, and number of passed courses. Display all results. A mark of 50 or higher is considered a pass.

```python
name = input("Enter student name: ")
m1 = float(input("Enter mark for course 1: "))
m2 = float(input("Enter mark for course 2: "))
m3 = float(input("Enter mark for course 3: "))
m4 = float(input("Enter mark for course 4: "))
m5 = float(input("Enter mark for course 5: "))

total = m1 + m2 + m3 + m4 + m5
average = total/5
highest = max(m1, m2, m3, m4, m5)
lowest = min(m1, m2, m3, m4, m5)

passed = 0

if m1 >= 50:
    passed = passed + 1
if m2 >= 50:
    passed = passed + 1
if m3 >= 50:
    passed = passed + 1
if m4 >= 50:
    passed = passed + 1
if m5 >= 50:
    passed = passed + 1

print("Student Name:", name)
print("Total Marks:", total)
print("Average Marks:", average)
print("Highest Mark:", highest)
print("Lowest Mark:", lowest)
print("Passed Courses:", passed)
```
