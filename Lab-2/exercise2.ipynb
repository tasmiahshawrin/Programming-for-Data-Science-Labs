1. Course Marks: Create a list containing the marks of five courses. Display the total, average, highest mark, and lowest mark. Count the number of passed courses, using 50 as the pass mark.

```python
marks = [77, 98, 53, 64, 92]

print("Marks:", marks)
print("Total marks:", sum(marks))
print("Average mark:", sum(marks) / len(marks))
print("Highest mark:", max(marks))
print("Lowest mark:", min(marks))

passed = 0
for mark in marks:
    if mark >= 50:
        passed += 1

print("Passed courses:", passed)
```

2. Weekend Temperatures: Create a tuple containing six temperature readings, three for Saturday and three for Sunday. Use a for loop to display each reading and calculate the highest and lowest temperatures.

```python
temperatures = (28, 29, 35, 30, 32, 27)

for temp in range(len(temperatures)):
    if temp < 3:
        print("Saturday:", temperatures[temp])
    else:
        print("Sunday:", temperatures[temp])

print("Highest temperature:", max(temperatures))
print("Lowest temperature:", min(temperatures))
```

3. Word Analyzer: Ask the user to enter a word or short sentence. Display the number of characters, vowels (a, e, i, o, u), and spaces. Use a for loop to examine each character.

```python
text = input("Enter a word or a short sentence: ")
characters = 0
vowels = 0
spaces = 0

for ch in text:
    characters += 1
    if ch.lower() in "aeiou":
        vowels += 1
    if ch == " ":
        spaces += 1

print("Characters:", characters)
print("Vowels:", vowels)
print("Spaces:", spaces)
```

4. Grade Calculator: Write a function named calculate_grade(marks) that receives a list of marks, calculates the average, and returns the corresponding grade.

80 or higher (A), 70 to below 80 (B), 60 to below 70 (C), 50 to below 60 (D), below 50 (F).

Call the function using a list of five marks.

```python
def calculate_grade(marks):
    avg = sum(marks) / len(marks)

    if avg >= 80:
        return "A"
    elif avg >= 70:
        return "B"
    elif avg >= 60:
        return "C"
    elif avg >= 50:
        return "D"
    else:
        return "F"

marks = [56, 75, 98, 82, 92]
print("Grade:", calculate_grade(marks))
print("Average:", sum(marks) / len(marks))
```

5. Shopping Cart Processor: Create a list of dictionaries representing four products. Each dictionary must contain a product name, price, and quantity.
Use a loop to calculate the total cost. Write a function named apply_discount(total) that applies a 10% discount when the total is at least 2,000. Otherwise, return the original total.

```python
products = [
    {"name": "Notebook", "price": 300, "quantity": 5},
    {"name": "Pencil", "price": 700, "quantity": 2},
    {"name": "Eraser", "price": 500, "quantity": 3},
    {"name": "Sharpner", "price": 400, "quantity": 1}
]
total = 0

for p in products:
    total += p["price"] * p["quantity"]

def apply_discount(total):
    if total >= 2000:
        return total * 0.90
    else:
        return total

print("Actual Cost:", total)
print("Cost after discount:", apply_discount(total))
```

6. Countdown: Write a program using a while loop that asks the user for a non-negative integer and prints a countdown from that number to zero.

```python
n = int(input("Enter a non-negative integer: "))

while n >= 0:
    print(n)
    n -= 1
```

7. Student Record Analyzer: Using the student records below, display each student's name and marks. Count the students who passed (marks >= 50), calculate the class average, and identify the student with the highest mark.

```python
students = [
    {"name": "Rahim", "department": "CSE", "marks": 85, "attendance": 92},
    {"name": "Karim", "department": "CSE", "marks": 48, "attendance": 76},
    {"name": "Sara", "department": "DS", "marks": 91, "attendance": 95},
    {"name": "Nadia", "department": "DS", "marks": 67, "attendance": 84},
    {"name": "Hasan", "department": "CSE", "marks": 73, "attendance": 78}
]

total = 0
passed = 0
highest = students[0]

for s in students:
    print(s["name"], s["marks"])
    total += s["marks"]
    if s["marks"] >= 50:
        passed += 1

    if s["marks"] > highest["marks"]:
        highest = s

average = total / len(students)

print("Passed:", passed)
print("Average Mark:", average)
print("Highest Mark:", highest["marks"])
print("Topper:", highest["name"])
```

8. Product Inventory Summary: Using the product records below, calculate each product's line total using price * quantity. Calculate the total inventory value and identify the product with the highest line total.

```python
products = [
    {"name": "Notebook", "price": 120, "quantity": 3},
    {"name": "Pen", "price": 20, "quantity": 5},
    {"name": "USB Drive", "price": 850, "quantity": 2},
    {"name": "Mouse", "price": 650, "quantity": 1}
]

total = 0
highest_total = 0
highest_product = ""

for product in products:
    line_total = product["price"] * product["quantity"]

    print(product["name"], line_total)

    total = total + line_total

    if line_total > highest_total:
        highest_total = line_total
        highest_product = product["name"]

print("Total Inventory Value:", total)
print("Product with Highest Line Total:", highest_product)
print("Highest Line Total:", highest_total)
```
